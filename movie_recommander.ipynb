```python
# merging both dataframes
movies = movies.merge(credits,on='title')  # Merge movies and credits DataFrames on the 'title' column
display("Merged DataFrame",movies.head(3)) # Display the first 3 rows of the merged DataFrame
movies.info()
```

```python
# imp column selection 
movies = movies[['movie_id','title','overview','genres','keywords','cast','crew']] # Extract relevant columns
movies.info()
movies.head(2)
```

```python
# Check for null values and handle them
movies.isnull().sum()       # Check for null values
movies.dropna(inplace=True) # Remove rows with null values
movies.isnull().sum()       # Verify that no null values remain
```

```python
# Check for duplicate rows and handle them
movies.duplicated().sum()             # Check for duplicate rows
movies.drop_duplicates(inplace=True)  # Remove duplicate rows
movies.duplicated().sum()             # Verify that no duplicate rows remain
```

```python
# Accessing specific data in the DataFrame in drifferent ways
movies.iloc[0,3]      # Access the 'genres' column of the first row if you know the index of the column. 
movies.iloc[0].genres # Access the 'genres' column of the first row if you don't know the index of the column.

movies['genres'][0]   # Access the 'genres' column of the first row without using iloc.
```

```python
# '[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}...]'
# extract only name 
# and apply on all dictionaries from original list because original list contains less ammount of dictionaries in 'genres' and 'keywords' columns.
# result: ["Action", "Adventure"...]

import ast

def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)      # Convert the 'genres' column to a list of genre names
```

```python
movies.head(2)                                          # confirm the changes made to the 'genres' column
```

```python
movies['keywords'] = movies['keywords'].apply(convert)  # Convert the 'keywords' column to a list of keyword names
```

```python
movies.head(2)                                          # confirm the changes made to the 'keywords' column
```

```python
# '[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}...]'
# extract only name 
# and apply on first 3 dictionaries from original list because original list contains huge ammount of dictionaries in 'cast' column.
# result: ["Action", "Adventure"...]

def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter < 3:  # Limit to the first 3 items
            L.append(i['name'])
            counter += 1
        else:
            break
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)          # Convert the 'cast' column to a list of the first 3 cast member names
```

```python
movies.head(2)                          # confirm the changes made to the 'cast' column
```

```python
# ['[{"credit_id": "52fe48009251416c750aca23", "department": "Director",...]'
# extract only deparrtment 

def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break  # Stop after finding the first director
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)     # Convert the 'crew' column to a list containing the director's name
```

```python
movies.head(2)                          # confirm the changes made to the 'crew' column
```

```python
movies.iloc[0].overview   # Access the 'overview' column of the first row we found it is a string and we will convert it to a list of words.
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split()) # Convert the 'overview' column to a list of words
```

```python
movies.head(2)                          # confirm the changes made to the 'overview' column
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])    # Remove spaces from genre names
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])# Remove spaces from keyword names
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])        # Remove spaces from cast names
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])        # Remove spaces from crew names
```

```python
movies.head(2)                          # confirm the changes made to the 'genres,keywords,cast,crew' column
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew'] # Combine all relevant columns into a single 'tags' column
```

```python
movies.head(2)                          # confirm the changes made to the 'tags' column
```

```python
new_df = movies[['movie_id','title','tags']]  # Create a new DataFrame with only the 'movie_id', 'title', and 'tags' columns
```

```python
new_df.head(2)  # Confirm the changes made to the new DataFrame
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x)) # Convert the 'tags' column from a list of words to a single string of words
```

```python
new_df.head(2)  # Confirm the changes made to the new DataFrame
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower()) # Convert the 'tags' column to lowercase for uniformity
```
