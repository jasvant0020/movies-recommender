⟶ 9. CONVERT GENRES AND KEYWORDS DATA
</p>

```python
# '[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}...]'
# extract only name 
# and apply on all dictionaries from original list because original list contains less ammount of dictionaries in 'genres' and 'keywords' columns.
# result: ["Action", "Adventure"...]

import ast

def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)      # Convert the 'genres' column to a list of genre names
```

```python
movies.head(2)                                          # confirm the changes made to the 'genres' column
```

```python
movies['keywords'] = movies['keywords'].apply(convert)  # Convert the 'keywords' column to a list of keyword names
```

```python
movies.head(2)                                          # confirm the changes made to the 'keywords' column
```

<p style="color:#075985; font-family:'Segoe UI', sans-serif; font-size:12px; font-weight:600; letter-spacing:2px; text-transform:uppercase; margin:20px 0 4px 0;">
⟶ 10. EXTRACT TOP 3 CAST MEMBERS
</p>

```python
# '[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}...]'
# extract only name 
# and apply on first 3 dictionaries from original list because original list contains huge ammount of dictionaries in 'cast' column.
# result: ["Action", "Adventure"...]

def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter < 3:  # Limit to the first 3 items
            L.append(i['name'])
            counter += 1
        else:
            break
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)          # Convert the 'cast' column to a list of the first 3 cast member names
```

```python
movies.head(2)                          # confirm the changes made to the 'cast' column
```

<p style="color:#075985; font-family:'Segoe UI', sans-serif; font-size:12px; font-weight:600; letter-spacing:2px; text-transform:uppercase; margin:20px 0 4px 0;">
⟶ 11. EXTRACT DIRECTOR FROM CREW
</p>

```python
# ['[{"credit_id": "52fe48009251416c750aca23", "department": "Director",...]'
# extract only deparrtment 

def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break  # Stop after finding the first director
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)     # Convert the 'crew' column to a list containing the director's name
```

```python
movies.head(2)                          # confirm the changes made to the 'crew' column
```

<p style="color:#075985; font-family:'Segoe UI', sans-serif; font-size:12px; font-weight:600; letter-spacing:2px; text-transform:uppercase; margin:20px 0 4px 0;">

```python
movies.head(2)                          # confirm the changes made to the 'overview' column
```

</p>

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])    # Remove spaces from genre names
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])# Remove spaces from keyword names
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])        # Remove spaces from cast names
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])        # Remove spaces from crew names
```

```python
movies.head(2)                          # confirm the changes made to the 'genres,keywords,cast,crew' column
```

<p style="color:#075985; font-family:'Segoe UI', sans-serif; font-size:12px; font-weight:600; letter-spacing:2px; text-transform:uppercase; margin:20px 0 4px 0;">
⟶ 18. APPLY STEMMING
</p>

```python
# ['loved','love','loves','loving']
# after applying stemming to reduce words to their root form. we will get only 'love' as a feature.
# ['love']

import nltk

from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))  # Apply stemming to each word in the text
    return " ".join(y)  # Join the stemmed words back into a single string
```

```python
new_df['tags'] = new_df['tags'].apply(stem) # Apply the stemming function to the 'tags' column
```

<p style="color:#075985; font-family:'Segoe UI', sans-serif; font-size:12px; font-weight:600; letter-spacing:2px; text-transform:uppercase; margin:20px 0 4px 0;">
⟶ 19. Vactorisation
</p>

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=5000,stop_words='english') # Create a CountVectorizer object to convert text data into numerical features
vector = cv.fit_transform(new_df['tags']).toarray() # Fit and transform the 'tags' column to create a feature matrix
```

```python
vector # Display the shape of the feature matrix
```

```python
list(cv.get_feature_names_out()) # Display the feature names
```

<p style="color:#075985; font-family:'Segoe UI', sans-serif; font-size:12px; font-weight:600; letter-spacing:2px; text-transform:uppercase; margin:20px 0 4px 0;">
⟶ 20. COSINE SIMILIARITY
</p>

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similiarity = cosine_similarity(vector) # Compute the cosine similarity between all pairs of movies based on their feature vectors
```

<p style="color:#075985; font-family:'Segoe UI', sans-serif; font-size:12px; font-weight:600; letter-spacing:2px; text-transform:uppercase; margin:20px 0 4px 0;">
⟶ 21. SORTING THE SIMILIARITY AND RECOMMENDATION FUNCTION
</p>

```python
sorted(list(enumerate(similiarity[0])),reverse=True,key=lambda x:x[1])[1:6] # Sort the movies based on similarity to the first movie and display the top 5 similar movies
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0] # Get the index of the movie in the DataFrame
    distances = similiarity[movie_index] # Get the similarity scores for the movie
    movies_list = sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6] # Sort the movies based on similarity and get the top 5
    for i in movies_list:
        print(new_df.iloc[i[0]].title) # Print the titles of the recommended movies
```

<p style="color:#075985; font-family:'Segoe UI', sans-serif; font-size:12px; font-weight:600; letter-spacing:2px; text-transform:uppercase; margin:20px 0 4px 0;">
⟶ 22. TEST THE MODEL 
</p>

```python
recommend('Batman Begins') # Call the recommend function to get movie recommendations based on 'Batman Begins'
```
